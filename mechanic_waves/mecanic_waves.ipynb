```python
import matplotlib.pyplot as plt
import numpy as np

# Your data
n = np.array([i for i in range(1, 7)])
freq = np.array([0.5, 0.8, 1.0, 1.2, 1.4, 1.7])
y_err = 0.1

# Perform linear fit (degree 1 polynomial)
slope, intercept = np.polyfit(n, freq, 1)
fit_line = slope * n + intercept

# Create the plot
plt.figure(figsize=(8, 6))

# Plot scatter chart with error bars
plt.errorbar(n, freq, yerr=y_err, fmt='o', label='Data with Error Bars', capsize=4)

# Plot the linear fit line
plt.plot(n, fit_line, 'r-', label=f'Linear Fit: y = {slope:.3f}x + {intercept:.3f}')

# Labeling
plt.xlabel('Number of Stationary Points')
plt.ylabel('freq [Hz]')
# plt.title('Scatter Chart with Linear Fit')
plt.legend()
plt.grid(True)

# Show the plot
plt.show()
```
